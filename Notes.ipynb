#### * Into to Python
- Python Program -> Interpreter -> Machine understable Language.
- Popular because: Simple, open source, High level, Portable.
- Compiler vs Interpreted: Compiler translates the entire program into machine code before execution.
- Interpreter: Translates and executes the program line by line during execution.
- High level vs Low level: Python, Java, C++ is a human readable programming language that is easy to write and understand.
- Low level: Assembly, Binary language is machine oriented programming language that is closer to hardware and harder for humans to understand.

#### 1. Variables.
- We give name to variable to identify the memory location.
- Data: value that stored in variable.
- Statically Typed: C, C++, Java where "int a" -> Store only integer "string a" -> Store only Characters.
- Python is Dynamically Types Language: In this no need to declare datatype just give value to variable explicitly and interpriter assumes it implicitly.
- a = 10, a = 8.20, a = "sai"
- Variable contains and start '_', abc, 123 but not any other spetial character and space like, 12abc - @a - $name - # ab
- Variables are case sensitive and python key words cannot used as variable.

#### 2. Comments.
- '#' Single line comment
- """ """, ''' ''' Multi Line
- sep: Separate multiple arguments in print one line. Arguments are called the variable passed in print().
- end: "" end with nothing, '_' end with space, "." end with dot

#### 3. Data Types.
- Tells the type of data stored in variable.
- Basic: Number- (int,float) Character- (str) Boolean- (bool: True '1', False '0') None- (Used when not sure about datatype of variable).
- Advanced: List, Tuple, Set, Dictionary, Class.
- Multiline Instructions:- \ used to indicate to interpreter that the next line is also the part of same instruction.

#### 4. Input Output
- Output: {} work as placeholder of value in print().
- This Formatted string will always start with 'f'.
- Input: One limitation of input() is that it takes input in form of String datatype, even the Number values.
