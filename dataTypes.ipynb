#### Integer (int)

```python
age = 22
print("Integer:", age)
print(type(age))
```

#### Float (float)

```python
percentage = 85.5
print("Float:", percentage)
print(type(percentage))
```

#### Complex (complex)

```python
number = 3 + 5j
print("\nComplex:", number)
print(type(number))
```

#### String (str)

```python
name = "Jhon"
print("\nString:", name)
print(type(name))
```

#### Boolean (bool)

```python
is_student = True
print("\nBoolean:", is_student)
print(type(is_student))
```

#### List (list)

```python
fruits = ["Apple", "Mango", "Banana", "Apple"]
print("\nList:", fruits)
print(type(fruits))
```

#### Tuple (tuple)

```python
colors = ("Red", "Green", "Blue", "Red")
print("\nTuple:", colors)
print(type(colors))
```

#### Set (set)

```python
numbers = {10, 20, 30, 10}
print("\nSet:", numbers)
print(type(numbers))
```

#### Dictionary (dict)

```python
student = {
    "name": "Yash",
    "age": 22,
    "course": "MCA"
}
print("\nDictionary:", student)
print(type(student))
```

#### Range (range)

```python
numbers_range = range(1, 6)
print("\nRange:", list(numbers_range))
print(type(numbers_range))
```

#### Bytes (bytes)

```python
data = b"Hello"
print("\nBytes:", data)
print(type(data))
```

#### Bytearray (bytearray)

```python
data_array = bytearray(5)
print("\nBytearray:", data_array)
print(type(data_array))
```

#### Frozenset (frozenset)

```python
frozen_numbers = frozenset({10, 20, 30})
print("\nFrozenset:", frozen_numbers)
print(type(frozen_numbers))
```

#### NoneType

```python
result = None
print("\nNoneType:", result)
print(type(result))
```
