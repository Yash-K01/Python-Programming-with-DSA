#### Data Type.

```python
# Interger (int)
age = 22
print("Integer:", age)
print(type(age))
```

```python
# Float (float)
percentage = 85.5
print("Float:", percentage)
print(type(percentage))
```

```python
# Complex (complex)
number = 3 + 5j
print("\nComplex:", number)
print(type(number))
```

```python
# String (str)
name = "Jhon"
print("\nString:", name)
print(type(name))
```

```python
# Boolean (bool)
is_student = True # Always capital T & F
print("\nBoolean:", is_student)
print(type(is_student))
```

```python
# List (list)
fruits = ["Apple", "Mango", "Banana", "Apple"]
print("\nList:", fruits)
print(type(fruits))
```

```python
# Tuple (tuple)
colors = ("Red", "Green", "Blue", "Red")
print("\nTuple:", colors)
print(type(colors))
```

```python
# Set (set)
numbers = {10, 20, 30, 10}
print("\nSet:", numbers)
print(type(numbers))
```

```python
# Dictionary (dict)
student = {
    "name": "Yash",
    "age": 22,
    "course": "MCA"
}
print("\nDictionary:", student)
print(type(student))
```

```python
# Range (range)
numbers_range = range(1, 6)
print("\nRange:", list(numbers_range))
print(type(numbers_range))
```

#### Bytes:
- A mutable object is one whose value can be changed after it is created.
- An immutable object is one whose value cannot be changed after it is created.

```python
# Bytes (bytes)
""" immutable binary data type used to store a sequence of bytes 
(values from 0 to 255). It is commonly used for binary data such 
as images, audio, videos, and files. """

data = b"PHello"
print(data[0])
print(type(data))
```

```python
# Bytearray (bytearray)
""" mutable binary data type used to store a sequence of bytes (values from 0 
to 255). Unlike bytes, its contents can be modified after creation. """

data_array = bytearray([65, 66, 67, 68])
print(data_array)
print(type(data_array))
```

#### Frozenset:
- Immutable version of a set. It stores unique (non-duplicate) elements, but cannot be modified after it is created.
- Duplicates are automatically removed.
- You cannot add or remove elements because frozenset is immutable.
- The order may vary because sets are unordered.

```python
# Frozenset (frozenset)
frozen_numbers = frozenset({40, 20, 30, 70, 100, 90})
print(frozen_numbers)
print(type(frozen_numbers))
```

```python
# NoneType (None)
result = None
print("\nNoneType:", result)
print(type(result))
```
