#### Python Variables

```python
#  name                 =                    Jhon 
# (variable name)   (Assignment Operator)  (Value assign) 
age = 22
print("Age:", age)
```

```python
percentage = 85.5
print("Percentage:", percentage)
```

```python
name = "Jhon"
print("Name:", name)
```

```python
is_student = True
print("Is Student:", is_student)
```

```python
a = 10
b = 20
sum_value = a + b
print("Sum:", sum_value)
```

```python
x, y, z = 10, 20, 30
print(x)
print(y)
print(z)
```

```python
p = q = r = 100
print(p, q, r)
```

```python
print(type(age))
print(type(percentage))
print(type(name))
print(type(is_student))
```
