#### Addition (+):
- Most of arithmetic operator will work with int, float, bool only.
- (+) works with string only when both values are string.

```python
a = 10
b = 20
print(a + b)
```

```python
s = 'Hii'
n = "Jon"
print(s +" "+ n)
```

```python
import builtins
k = 5
print(s + builtins.str(k))
```

#### Subtraction (-):

```python
print(b - a)
```

#### Multipication (*):

```python
print(a * b)
```

```python
str = "Hii"
print(str * 5) # User string with integer multiplication for repetition.
```

#### Division (/):

```python
print(b / a)
```

```python
c = 0
print(c / a) # The zero divide by non zero number works.
```

```python
print(a / c) # The non zero divide by zero is mathematically undefined and will throw an error.
```

#### Modulus (%):

```python
print(b % a) # The 20 divided by 10 gives 0 remainder. Modulus gives the remainder of the division.
```

#### Integer Division (//):

```python
a = 35
b = 2
print(a / b)
print(a // b) # // Floor division gives the quotient without fractional part just gives the integer part of the quotient.
```

#### Exponential (**):

```python
a = 2
b = 2
print(a ** b) # Gives the power of a number.
```
